### Missing Values

```python
df.isna().sum() # check missing values
```

```python
# drop missing values
df = df.dropna() # as long as there is one missing value, the row will be dropped
df = df.dropna(axis = 1) # drop columns with missing values


df = df.dropna(subset = ["airline"]) # drop rows where "airline" is missing
df = df.dropna(subset = ["airline", "flight"]) # drop rows where "airline" or "flight" is missing
```

```python
# filling missing values
df = df.fillna(0) # fill missing values with 0
df = df.fillna(method = "ffill") # fill missing values with the previous value
df = df.fillna(method = "bfill") # fill missing values with the next value
```

```python
# replace missing values with a specific value
df = df.replace(np.nan, 0) # replace missing values with 0
df["airline"] = df["airline"].replace("AirAsia", "Air Asia")
```

```python
# drop duplicates
df = df.drop_duplicates(subset = ['airline', 'flight'], keep = 'first')
# keep = 'first'
# keep = 'last'
# keep = False -> drop all duplicates
```

# Group By

Common Aggregation Functions  
1. max()/min()
2. mean()/median()
3. sum()
4. count()/size() -> count() ignores NULL, while size() does not.
5. std()/var()
6. nunique() # count the number of unique values

```python
new_series = df.groupby('Students')['Scores'].sum()
new_series
```

```python
new_df = df.groupby('Students', as_index = False)['Scores'].sum()
new_df
```

```python
## multiple 
new_df2 = df.groupby('Students', as_index = False)['Scores'].agg(['mean', 'sum','max', 'min'])
new_df2
```

```python
## multiple 
new_df3 = df.groupby('Students', as_index = False).agg({'Courses':'count', 'Scores':'sum'})
new_df3
```

```python
## keep original rows -> transforms!!!
df['sum_of_scores'] = df.groupby('Students', as_index = False)['Scores'].transform('sum')
df
```

**Ranking**  
Use with groupby.  
rank(method = 'method', ascending = True/False)  
method can be:  
1. average: 重复则取平均名次
2. min: 同一组重复值取最小的名次，相当于RANK
3. max: 同一组重复值取最大的名次
4. first: 同一组重复值按出现顺序分配名词，相当于row_number
5. dense: 同一组重复值取最小的名词，且不留空缺，相当于dense_rank

```python
### Ranking
df['rank'] = df.groupby('Students')['Scores'].rank(method = 'min', ascending = False)
df
```

```python
## rolling window -> if sorting needed, should sort at first
df['rolling_mean'] = df.groupby('Students')['Scores'].transform(lambda x: x.rolling(
    window = 3, min_periods = 3).mean()
    )
df
```

```python
## cumulative sum
df['cum_sum'] = df.groupby('Students', as_index = False)['Scores'].transform(lambda x: x.expanding().sum())
df
```

```python
### shift
df['shifted_scores'] = df['Scores'].shift(1) # shift down
df['shifted2_scores'] = df['Scores'].shift(-1) # shift up
df
```

```python
## shift with group by
df['shifted3_scores'] = df.groupby('Students')['Scores'].transform(lambda x: x.shift(1))
df
```
